观察我设立的电力数据npz数据集里面是什么格式

```python
import numpy as np

data = np.load("electricityLondon/electricityLondon.npz")

print(data.files)
```

```python
for key in data.files:
    arr = data[key]
    print(f"Key: {key}")
    print(f"  Shape: {arr.shape}")
    print(f"  Dtype: {arr.dtype}")
    print("-" * 40)
```

```python
key = data.files[0]
arr = data[key]

print(arr[0])          # 第一个样本
print(arr[0].shape)    # 再确认一次维度
```
