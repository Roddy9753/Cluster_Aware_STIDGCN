观察原本的npz数据集里面是什么格式

## 观察test

```python
import numpy as np

data = np.load("PEMS08/test.npz")

print(data.files)
```

```python
for key in data.files:
    arr = data[key]
    print(f"Key: {key}")
    print(f"  Shape: {arr.shape}")
    print(f"  Dtype: {arr.dtype}")
    print("-" * 40)
```

```python
key = data.files[0]
arr = data[key]

print(arr[0])          # 第一个样本
print(arr[0].shape)    # 再确认一次维度
```

```python
#看一眼"x_offsets" "y_offsets"内部是什么

import numpy as np

data = np.load("PEMS08/test.npz")

x_offsets = data["x_offsets"]
y_offsets = data["y_offsets"]

print("x_offsets shape:", x_offsets.shape)
print("y_offsets shape:", y_offsets.shape)

print("\nRaw x_offsets:\n", x_offsets)
print("\nRaw y_offsets:\n", y_offsets)

# 转成一维方便看
print("\nFlattened:")
print("x_offsets:", x_offsets.flatten())
print("y_offsets:", y_offsets.flatten())
```
