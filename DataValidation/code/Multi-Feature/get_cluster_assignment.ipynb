```python
import pandas as pd

# 读取csv文件
df = pd.read_csv("gmm_probabilities_K3.csv", index_col=0)

# 获取每个节点概率最大的cluster编号
# idxmax会返回列名，例如 Cluster_2
max_clusters = df.idxmax(axis=1)

# 提取cluster编号（去掉"Cluster_"）
node_cluster_list = [int(c.split("_")[1]) for c in max_clusters]

# 输出结果
print("node_cluster_list:", node_cluster_list)

# 输出list维度
print("list的维度: 1 ×", len(node_cluster_list))
```
